# Sarcasm Detection Exploration

In this notebook, we will perform exploratory data analysis (EDA) on the sarcasm detection dataset. We will analyze the data distribution, visualize the text data, and prepare it for model training.

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load the dataset
data_path = '../data/processed/sarcasm_data.csv'  # Update with the correct path
df = pd.read_csv(data_path)

# Display the first few rows of the dataset
df.head()
```

```python
# Check the distribution of labels
plt.figure(figsize=(8, 6))
sns.countplot(x='label', data=df)
plt.title('Distribution of Sarcasm Labels')
plt.xlabel('Label (0: Not Sarcastic, 1: Sarcastic)')
plt.ylabel('Count')
plt.show()
```

```python
# Visualize some examples of sarcastic and non-sarcastic comments
sarcastic_samples = df[df['label'] == 1].sample(5)
non_sarcastic_samples = df[df['label'] == 0].sample(5)

print('Sarcastic Samples:')
print(sarcastic_samples['comment'].values)

print('\nNon-Sarcastic Samples:')
print(non_sarcastic_samples['comment'].values)
```

## Text Preprocessing

Next, we will preprocess the text data to prepare it for model training. This includes cleaning the text, tokenizing, and padding sequences.

```python
# Example of text cleaning function
import re

def clean_text(text):
    text = re.sub(r'http\S+|www\S+', '', text)  # Remove URLs
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)  # Remove punctuation
    text = text.lower()  # Convert to lowercase
    return text

# Apply text cleaning
df['cleaned_comment'] = df['comment'].apply(clean_text)
df.head()
```
